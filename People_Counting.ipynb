```python
from ultralytics import YOLO
```

```python
model=YOLO("yolo11n.pt")
```

```python
%%writefile botsort_reid.yaml
tracker_type: botsort
track_high_thresh: 0.1
track_low_thresh: 0.05
new_track_thresh: 0.1
track_buffer: 60
match_thresh: 0.7
fuse_score: True

# Camera motion compensation
gmc_method: sparseOptFlow

# Re-Identification (ReID) enabled
proximity_thresh: 0.5
appearance_thresh: 0.25
with_reid: True
model: auto
```

```python
unique_ids = set()

results = model.track(
    source="test1.mp4",
    classes=[0],                   # 0 = person class only
    conf=0.60,
    tracker="botsort_reid.yaml",
    persist=True,
    save=True,
    stream=True
)

for r in results:
    if r.boxes.id is not None:
        ids = r.boxes.id.int().cpu().tolist()
        unique_ids.update(ids)

print(f"Total unique people counted: {len(unique_ids)}")
```
